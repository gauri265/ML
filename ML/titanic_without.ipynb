```python

import math
```

```python
X = [
    [3,0,22],
    [1,1,38],
    [3,1,26],
    [1,1,35],
    [3,0,35],
    [3,0,54],
    [1,0,2],
    [3,1,27],
    [2,1,14],
    [3,0,20]
]

y = [0,1,1,1,0,0,1,1,1,0]

n=len(X)
m=len(X[0])
```

```python
weights = [0.0] * m
bias = 0.0

learning_rate=0.01
epochs=1000
```

```python
def sigmoid(z):
    return 1/(1+math.exp(-z))
```

```python
for epoch in range(epochs):
    dw=[0.0]*m
    db=0.0
    for i in range(n):
        z=bias
        for j in range(m):
            z=z+weights[j] * X[i][j]
        pred=sigmoid(z)
        error=pred-y[i]
        for j in range(m):
            dw[j]=dw[j]+error*X[i][j]
        db=db+error
    for j in range(m):
        weights[j]=weights[j]-learning_rate*dw[j]/n
    bias=bias-learning_rate*db/n
```

```python
print("Weights: ",weights)
print("Bias: ",bias)
def predict(sample):
    z=bias
    for j in range(m):
        z=z+weights[j]*sample[j]
    p=sigmoid(z)
    if p>=0.5:
        return 1
    else:
        return 0
    
```

```python
correct = 0
for i in range(n):
    pred = predict(X[i])
    if pred == y[i]:
        correct += 1

accuracy = (correct / n) * 100

print("Accuracy:", round(accuracy, 2), "%")
print("Correct predictions: ",correct)
```

```python
new_passenger = [2, 1, 30]

prediction = predict(new_passenger)

if prediction == 1:
    print("Prediction: Survived")
else:
    print("Prediction: Did Not Survive")
```
